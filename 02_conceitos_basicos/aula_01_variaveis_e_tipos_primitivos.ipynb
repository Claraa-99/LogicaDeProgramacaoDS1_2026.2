### 🔍 O Cenário da Investigação (Missão Casas Paulino)
Nossa escola assumiu um projeto para digitalizar a gestão de clientes e anúncios da loja **Casas Paulino** em Tianguá. Precisamos cadastrar:
* Nome do cliente
* Idade
* Valor investido em compras
* Se o cadastro está ativo ou inativo


```python
# Exemplo do Professor: Criando variáveis para o cliente
nome_cliente = "Maria das Dores"       # str
idade_cliente = 28                    # int
valor_compras = 249.90                # float
cadastro_ativo = True                 # bool

# Inspecionando os tipos armazenados na memória
print("Tipo do nome:", type(nome_cliente))
print("Tipo da idade:", type(idade_cliente))
print("Tipo do valor:", type(valor_compras))
print("Tipo do cadastro:", type(cadastro_ativo))
```

## 2. Regras de Nomenclatura no Python (PEP 8)
* Use nomes descritivos em letras minúsculas separadas por sublinhado: `snake_case` (ex: `valor_total`, `taxa_servico`).
* **Não pode** começar com números (ex: `1nota` ❌).
* **Não pode** conter espaços (ex: `nome cliente` ❌).
* **Não pode** conter caracteres especiais como acentos ou cedilha (ex: `preço` ❌ -> use `preco` ✅).
* Não use palavras reservadas do Python (como `if`, `for`, `print`, `class`).
